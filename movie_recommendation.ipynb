### Import Necessary Libraries

```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics.pairwise import cosine_similarity
import warnings
warnings.filterwarnings('ignore')
```

### Feature Processing

```python
df['genres'] = df['genres'].str.split('|').str.join('').str.lower()

# Converting the genres column to vector
vectorizer = TfidfVectorizer()
genre_vector = vectorizer.fit_transform(df['genres'])
```

### Building the Recommendation System

```python
def recommendation(title = 'Grumpier Old Men (1995)', top_n = 5):
    # Getting the index of the movie
    idex = df[df['title'] == title].index[0]
    
    # Calculating the similarity of the targeted movie
    similarity = cosine_similarity(genre_vector[idex], genre_vector).flatten()
    
    # Pairing the similarity score with index and sort by scores
    score = list(enumerate(similarity))
    score.sort(key = lambda x: x[1], reverse = True)
    
    # Selecting top_n movies index
    movie_index = [i[0] for i in score[1: top_n + 1]]
    return df[['title']].iloc[movie_index].reset_index(drop = True)
    
    
    





recommendation()    
```

```python
recommendation('Sabrina (1995)') 
```
